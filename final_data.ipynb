```python
import pandas as pd

# Load CSV
df = pd.read_csv("datasets/data_2020_2023.csv")

# Step 1: Drop columns with >50% missing values
df = df[df.columns[df.isnull().mean() < 0.5]]

# Step 2: Fill missing numerical values
numeric_cols = df.select_dtypes(include='number').columns
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors=3)
df[numeric_cols] = pd.DataFrame(imputer.fit_transform(df[numeric_cols]), columns=numeric_cols)

# Step 3: Fill missing categorical values (if any)
categorical_cols = df.select_dtypes(include='object').columns
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Check result
print("Missing values after preprocessing:")
print(df.isnull().sum())
```

```python
# Save the preprocessed DataFrame to a new CSV
df.to_csv("datasets/data_2020_2023_preprocessed.csv", index=False)

print("Preprocessed dataset saved as 'data_2020_2023_preprocessed.csv'")
```
