```python
df2 = pd.read_csv("datasets/country_yearly_covid_summary.csv")
```

```python
df2['Year'] = pd.to_numeric(df2['Year'], errors='coerce').astype('Int64')
years = [2017, 2018, 2019, 2020]
countries = df2['Country'].unique()

new_rows = []
for c in countries:
    for y in years:
        if not ((df2['Country'] == c) & (df2['Year'] == y)).any():
            new_rows.append({'Country': c, 'Year': y, 'Total_cases': 0.0, 'Total_deaths': 0.0})

if new_rows:
    df2 = pd.concat([df2, pd.DataFrame(new_rows)], ignore_index=True)
    df2 = df2.sort_values(['Country', 'Year']).reset_index(drop=True)

df2.to_csv("datasets/country_yearly_covid_summary_with_2017_2020.csv", index=False)
```

```python
import pandas as pd

# Read the CSV files
covid_df = pd.read_csv('datasets/country_yearly_covid_summary_with_2017_2020.csv')
economic_df = pd.read_csv('datasets/data_2017_2025.csv')

# Display initial info
print(f"COVID dataset shape: {covid_df.shape}")
print(f"Economic dataset shape: {economic_df.shape}")

# Perform inner join to keep only matching countries and years
merged_df = pd.merge(
    covid_df,
    economic_df,
    left_on=['Country', 'Year'],
    right_on=['country_name', 'year'],
    how='inner'
)

# Drop duplicate columns (country_name and year since we have Country and Year)
merged_df = merged_df.drop(['country_name', 'year'], axis=1)

# Display merge results
print(f"\nMerged dataset shape: {merged_df.shape}")
print(f"\nFirst few rows of merged dataset:")
print(merged_df.head())

# Save the merged dataset
merged_df.to_csv('datasets/merged_2.csv', index=False)
print("\nMerged dataset saved to 'datasets/merged_2.csv'")

# Show which countries are in the final dataset
merged_df.head(10)
```

```python
# Load the CSV, treating empty strings as NaN
df = pd.read_csv("datasets/merged.csv", na_values=["", " "])

# Drop rows where 'GDP Growth (% Annual)' is missing
df = df.dropna(subset=["GDP Growth (% Annual)"])

# Overwrite the original CSV
df.to_csv("datasets/merged.csv", index=False)
```
